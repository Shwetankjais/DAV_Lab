```python
import pandas as pd
import numpy as np
```

```python
data = {
    'Name':['Amit','Bebo','Chetan','Amit','Divya','Ravi'],
    'Age' :[25, np.nan, 30,25,28,22],
    'Marks':[85,90, np.nan, 85,78,95],
    'City':['Mysore','Mysore','Bangalore','Mysore',np.nan,'Delhi']
}

df=pd.DataFrame(data)
print("Datasets:")
print(df)
```

```python
print("\n Shape of Datasets (rows,columns):",df.shape)
print("\n Columns names:",df.columns.tolist())
print("\n Dta types of each columns:")
print(df.dtypes)
print("\n Dataset info:")
df.info()
```

```python
print("\n Descriptive statistics (numerical,columns):")
print(df.describe())

print("\n Mean:\n ",df.mean(numeric_only=True))
print("\n Median:\n ",df.median(numeric_only=True))
print("\n Standard Deviation:\n ",df.std(numeric_only=True))
```

```python
print("\n Missing values in each columns:")
print(df.isnull().sum())
print("\n Total missing values in dataset:", df.isnull().sum().sum())
```

```python
print("\n Duplicate rows in datasets:")
print(df[df.duplicated()])
print("\n Number of duplicate rows:",df.duplicated().sum())
print("\n EDA process completed successfully!")
```
